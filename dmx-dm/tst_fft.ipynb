```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.fft as fft
```

```python
npix = 34
npts = 2 ** 14
A = 1
fs = 100
```

```python
t = np.linspace(0, 1, npts)
y = A * np.sin(2 * np.pi * fs * t) + np.random.randn(npts)
```

```python
# Analyse spectrale avec rfft
f = np.fft.rfftfreq(npts, 1 / fs)  # Fréquences associées
power_spectrum = abs(fft.rfft(y))

plt.loglog(f, power_spectrum)
```

```python
y34 = y[::34]
npts34 = len(y34)
fs34 = fs / 34

# Analyse spectrale avec rfft
f34 = np.fft.rfftfreq(npts34, 1 / fs34)  # Fréquences associées
power_spectrum34 = abs(fft.rfft(y34))

plt.loglog(f34, power_spectrum34)
```
